```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, LogisticRegression, RidgeClassifier, Lasso,Ridge
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
data = pd.read_csv("heart.csv")
data
x = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
x_train , x_test,y_train ,y_test = train_test_split(x,y ,test_size=0.2,random_state=42)
```

```python
param = {"alpha":[1,2,3,0.5,4],
        "tol":[0.001,0.0001,0.00001,0.000001]}
RC_grid = GridSearchCV(
    estimator=RidgeClassifier(),
    param_grid=param,
    scoring='accuracy',
    cv=5)
```

```python
RC_grid_pipline = Pipeline([("Grid Search Cv",RC_grid)])
RC_grid_pipline.fit(x_train,y_train)
```

```python
y_predict = RC_grid_pipline.predict(x_test)
y_predict
```

```python
accuracy = accuracy_score(y_test, y_predict)
precision=precision_score(y_test, y_predict)
recall=recall_score(y_test, y_predict)
f1=f1_score(y_test, y_predict)
print("accuracy",accuracy)
print("f1",f1)
print("recall",recall)
print("precision", precision)
```

```python
data1 =[accuracy,f1,recall,precision]
logistic_re= pd.DataFrame(data1,index=["accuracy","f1","recall","Precision"],columns=["ridgeclassifier"])
```

```python
lasso_param = {"C": [0.01, 0.1, 1, 10, 100],
    "tol": [0.001, 0.0001, 0.00001, 0.000001]}
lasso_RC_grid = GridSearchCV(
    estimator= LogisticRegression(
        penalty="l1",
        solver="liblinear",
        max_iter=1000),
    param_grid=lasso_param,
    scoring='accuracy',
    cv=5)
```

```python
lasso_RC_grid_pipline = Pipeline([("Grid Search Cv",lasso_RC_grid)])
lasso_RC_grid_pipline.fit(x_train,y_train)
```

```python
lasso_y_predict = lasso_RC_grid_pipline.predict(x_test)
lasso_accuracy = accuracy_score(y_test, lasso_y_predict)
lasso_precision=precision_score(y_test, lasso_y_predict)
lasso_recall=recall_score(y_test, lasso_y_predict)
lasso_f1=f1_score(y_test, lasso_y_predict)
```

```python
lasso_data= [lasso_accuracy,lasso_f1,lasso_recall,lasso_precision]
logistic_lasso= pd.DataFrame(lasso_data,index=["accuracy","f1","recall","Precision"],columns=["lassoclassifier"])
logisticregression = pd.concat([logistic_lasso, logistic_re],axis=1)
logisticregression
```

```python
std = pd.read_csv("student_exam_performance.csv")
std.head()
```

```python
std.isna().sum()
std.dropna(inplace=True)
std.isna().sum()
```

```python
x= std.drop(["exam_score","student_id", "gender", "education_level", "school_type", "family_income", "parent_education", "urban_rural", "class_participation", "study_consistency", "study_environment", "study_method", "revision_frequency", "notes_quality", "sleep_quality", "break_frequency", "motivation_level", "device_availability", "educational_app_usage", "exam_difficulty", "performance_grade", "pass_status", "performance_level"], axis=1)
y = std["exam_score"]
```

```python
x_train , x_test,y_train ,y_test = train_test_split(x,y ,test_size=0.3,random_state=42)
```

```python
param = {"alpha":[1,2,3,0.5,4],
        "tol":[0.001,0.0001,0.00001,0.000001]}
RC_grid = GridSearchCV(
    estimator=Ridge(),
    param_grid=param,
    scoring='r2',
    cv=5)
```

```python
RC_grid_pipline = Pipeline([("Grid Search Cv",RC_grid)])
RC_grid_pipline.fit(x_train,y_train)
```

```python
r_y_test_pred = RC_grid_pipline.predict(x_test)
```

```python
r_MAE =mean_absolute_error(y_test,r_y_test_pred)
r_RMSE= root_mean_squared_error(y_test,r_y_test_pred)
r_R2_Score =r2_score(y_test,r_y_test_pred)
```

```python
lasso_param = {
    "alpha": [0.01, 0.1, 1, 10, 100],
    "tol": [0.001, 0.0001, 0.00001, 0.000001]}
lasso_RC_grid = GridSearchCV(
    estimator=Lasso(),
    param_grid=lasso_param,
    scoring='r2',
    cv=5)
```

```python
lasso_RC_grid_pipline = Pipeline([("Grid Search Cv",lasso_RC_grid)])
lasso_RC_grid_pipline.fit(x_train,y_train)
```

```python
l_y_test_pred = lasso_RC_grid_pipline.predict(x_test) 
```

```python
l_MAE =mean_absolute_error(y_test,l_y_test_pred)
l_RMSE= root_mean_squared_error(y_test,l_y_test_pred)
l_R2_Score =r2_score(y_test,l_y_test_pred)
```

```python
l_l = [l_MAE,l_RMSE,l_R2_Score]
df_l_l = pd.DataFrame(l_l,index= ["MAE","RMSE","R2"],columns=["lassoclassifier"])
```

```python
l_r = [r_MAE,r_RMSE,r_R2_Score]
df_l_r = pd.DataFrame(l_r,index= ["MAE","RMSE","R2"],columns=["ridgeclassifier"])
df_l_r
```

```python
linearregression = pd.concat([df_l_r, df_l_l],axis=1)
linearregression 
```
